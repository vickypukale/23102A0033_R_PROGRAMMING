# **Loop - For Loop**

```r
for(i in 1:5)
    print(i^2)

# print("===============================================================")
cat(rep("=",25),"\n")


for(i in c(2,4,6,7))
  print(i^2)

cat(rep("=",25),"\n")

x = c(2,4,6,8,10,12)
excount = function(x){
    count = 0
    for(xval in x){
        if(xval / 2 > 3)
          count = count+1
    }
    print(count)
}
excount(x)

cat(rep("=",25),"\n")

child = c("child1" , 'child2' , "child3")
sweet = c("sweet1" , "sweet2", "sweet3")

for(x in child){
    for(y in sweet){
        print(paste(x,y))
    }
}

cat(rep("=",25),"\n")

drink = c("coffee" , "lemonade" , "tea", "juice")
for(x in drink){
    if(x == "tea"){
        break
    }
    print(x)
}
cat(rep("=",25),"\n")

for(x in drink){
    if(x == "lemonade"){
        next
    }
    print(x)
}
cat(rep("=",25),"\n")
for(x in drink){
    if(x == "tea"){
        next
    }
    print(x)
}
cat(rep("=",25),"\n")
```

# **Loop - While and repeat**

```r
i = 1
while(i < 10){
  print(i^2)
  i = i +2
}

cat(rep("=",25),"\n")

sumfuntion = function(){
  sum =0
  number = as.integer(readline(prompt= "Please slelct any number less than 25: "))
  while(number <= 25){
    sum = sum +number
    number = number +1
  }
  print(paste("The sum of number received from the While loop : ",sum))
}

# sumfuntion()
# 22 out = 94

cat(rep("=",25),"\n")

i = 1
repeat{
  print(i^2)
  i = i + 2
  if(i > 10 )
    break
}
cat(rep("=",25),"\n")

i = 1
repeat{
  i = i +1
  if(i < 10) next
  print(i^2)
  if(i >= 13)
    break
}
cat(rep("=",25),"\n")

```

# **Functions**

```r
abc = function(x){
  x^2
}
abc(3)
abc(6)
abc(9)
cat(rep("=",25),"\n")

abc = function(x,y){
  x^2+y^2
}

abc(3,4)
abc(10,10)
abc(-2,-3)

cat(rep("=",25),"\n")

abc = function(){
  for(i in 1:3){
    print(i^3)
  }
}
abc()
cat(rep("=",25),"\n")

```

# **Sequences**

```r
seq(from = 2 , to = 4)
cat(rep("=",25),"\n")
seq(from = 4 , to = 2)
cat(rep("=",25),"\n")
seq(from = -4 , to = 4)

cat(rep("=",25),"\n")

seq(from = 10 , to = 20 , by = 2)
cat(rep("=",25),"\n")

seq(from = 20 , to = 10 , by = -2)
cat(rep("=",25),"\n")

seq(from = 3 , to = -2 , by = -0.5)
cat(rep("=",25),"\n")

seq(to = -1 , length = 10)
cat(rep("=",25),"\n")

seq(from = 10 , length = 10)
cat(rep("=",25),"\n")

seq(from = 10 , length = 10 , by = 0.1)
cat(rep("=",25),"\n")

seq(from = 10 , length = 10 , by = -2)
cat(rep("=",25),"\n")

seq(from = 10 , length = 5 , by = -.2)
cat(rep("=",25),"\n")
```
