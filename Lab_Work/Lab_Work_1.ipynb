# R Operators and R Data Types

```r
a <- 10
b <- 3
print(a + b) # 13
print(a - b) # 7
print(a * b) # 30
print(a / b) # 3.333//floating point division
print(a ^ b) # 1000
print(a %% b) # 1 (remainder of integer division)
print(a %/% b) # 3 (quotient of integer division)
```

```r
x <- 10
y <- 5
print(x == y) # FALSE
print(x != y) # TRUE
print(x != y) # TRUE
print(x > y) # TRUE
print(x < y) # FALSE
print(x >= y) # TRUE
print(x <= y) # FALSE
```

```r
a <- TRUE
b <- FALSE
print(a & b) # FALSE
print(a | b) # TRUE
print(!a) # FALSE
print(a && b) # FALSE (stops if first condition is FALSE)
print(a || b) # TRUE (stops if first condition is TRUE)
```

```r
# Sequence operator
seq <- 1:5
print(seq) # 1 2 3 4 5
# %in% operator
print(3 %in% c(1, 2, 3)) # TRUE
# Matrix multiplication
mat1 <- matrix(1:4, nrow=2)
mat2 <- matrix(1:25,nrow=5)
# print(mat1 %*% mat2)
# Output:
# [,1] [,2]
# [1,] 19 22
# [2,] 43 50

m <- matrix(20:1, nrow = 4 , ncol = 5)
print(m %*% mat2)

#      [,1] [,2] [,3] [,4] [,5]
# [1,]  140  440  740 1040 1340
# [2,]  125  400  675  950 1225
# [3,]  110  360  610  860 1110
# [4,]   95  320  545  770  995

print(m[3][3])
```

```r
num <- 10.5 # Numeric
int <- 5L # Integer
char <- "Hello" # Character
bool <- TRUE # Logical
comp <- 3 + 2i # Complex
print(class(num)) # "numeric"
print(class(int)) # "integer"
print(class(char)) # "character"
print(class(bool)) # "logical"
print(class(comp)) # "complex"
```
