Operation on string : Substitution

```r
y = "Number of participants: 25"
sub("25","30",y)

z = "Mr. Singh is the smart one. Mr. Singh is funny , too."
z
sub("Mr. Singh","Professor Jha",z) # sub only first


z = "Mr. Singh is the smart one. Mr. Singh is funny , too."
z
gsub("Mr. Singh","Professor Jha",z) # sub all the words
```

Operation with strings: grep

```r
str = c("R Course","exercises","include examples of R language")
grep("ex",str,value=T)

str = c("R Course","exercises","include examples of R language")
grep("ex",str,value=F)

str = c("R Course","exercises","include examples of r language","R language")
grep("R",str,ignore.case = F ,value=T)
grep("R",str,ignore.case = T ,value=T)
grep("R",str,ignore.case = F ,value=F)
grep("R",str,ignore.case = T ,value=F)

x = "R course 24.07.2021"
y = "Number of participants: 25"
c(x,y)
grep("our",c(x,y))

grep("Num",c(x,y))
```
